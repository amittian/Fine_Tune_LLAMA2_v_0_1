```python
def _map_topics_to_embeddings(
        self, embeddings: np.ndarray, topics: List[int]
    ) -> np.ndarray:
        """
        Reduce the dimensionality of embeddings to 2D and map topics to the embeddings.

        Arguments:
            embeddings: The embeddings to use
            topics: The topics to use

        Returns:
            embeddings: The 2D embeddings
        """
        if self.topic_embeddings_ is not None:
            embeddings = self.topic_embeddings_
        elif isinstance(embeddings, np.ndarray):
            embeddings = self._reduce_dimensionality(embeddings)
        else:
            raise ValueError(
                "Please provide topic embeddings or specify an embedding model"
            )

        indices = [int(topic.item()) if isinstance(topic, np.float32) else int(topic) for topic in topics]
        embeddings = embeddings[indices]
        return embeddings
```
